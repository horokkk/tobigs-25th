# 과제1

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN

```

#### 데이터 생성

```python
X, y = make_moons(
    n_samples=200,
    noise=0.05,
    shuffle=True,
    random_state=0
)
```

#### 원본 데이터 시각화

```python
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], s=15, edgecolor='k')
plt.title("Original Moon-shaped Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

### K-Means 기본 클러스터링 수행

```python
kmeans_basic = KMeans(n_clusters=4, random_state=0)
labels_basic = kmeans_basic.fit_predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels_basic, cmap='viridis', s=15)
plt.title("K-Means Clustering (K=4)")
plt.show()
```

### Elbow Method를 통한 최적 클러스터 개수 탐색

```python
inertia_list = []
k_candidates = range(1, 15)

for k in k_candidates:
    km = KMeans(n_clusters=k, random_state=0)
    km.fit(X)
    inertia_list.append(km.inertia_)

plt.figure(figsize=(7, 5))
plt.plot(k_candidates, inertia_list, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

### 선택된 K값 기반 최종 클러스터링 결과 시각화

문제 1)  Elbow Method 결과를 참고하여 아래 코드의 빈칸에 가장 적절한 클러스터 개수 K를 입력하세요.

```python
optimal_k = 2 # Elbow Method에서 관성 감소가 완만해지는 지점

kmeans_final = KMeans(n_clusters=optimal_k, random_state=0)
labels_kmeans = kmeans_final.fit_predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels_kmeans, cmap='viridis', s=15)
plt.scatter(
    kmeans_final.cluster_centers_[:, 0],
    kmeans_final.cluster_centers_[:, 1],
    c='red', s=120, marker='X', label='Centroids'
)
plt.title(f"K-Means Result (K={optimal_k})")
plt.legend()
plt.show()
```

### DBSCAN 클러스터링 결과 시각화

```python
dbscan_model = DBSCAN(eps=0.3, min_samples=5)
labels_dbscan = dbscan_model.fit_predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels_dbscan, cmap='viridis', s=15)
plt.title("DBSCAN Clustering Result")
plt.show()
```

```python
print("K-Means cluster count:", len(np.unique(labels_kmeans)))
print("DBSCAN cluster count (excluding noise):", len(set(labels_dbscan)) - (1 if -1 in labels_dbscan else 0))
print("DBSCAN noise points:", np.sum(labels_dbscan == -1))
```

문제 2) 알고리즘 적합성 판단
K-Means와 DBSCAN의 클러스터링 결과를 비교하세요. 두 결과 중 본 데이터 구조에 더 적합하다고 판단되는 알고리즘을 하나 선택하고 그 이유를 데이터의 형태와 알고리즘의 특성을 연결하여 설명하세요.

답변 :
- K-Means가 부적합한 이유:
강의 자료에서 언급했듯이, K-Means는 "원 모양 클러스터를 추구" 하고 "지역적 패턴이 있는 군집 판별에 어려움" 이 있습니다. 반달 형태처럼 비선형적으로 굽어진 데이터는 centroid 기반의 거리 계산으로는 올바르게 분리할 수 없어, 두 반달이 뒤섞인 잘못된 클러스터링 결과가 나옵니다.

- DBSCAN이 적합한 이유:
반면 DBSCAN은 밀도 기반 알고리즘으로, 일정 반경(ε) 안에 충분한 이웃이 모인 고밀도 영역을 하나의 군집으로 확장합니다. 반달 모양처럼 복잡한 형상의 군집도 밀도를 따라 자연스럽게 연결되기 때문에 정확하게 두 개의 클러스터를 찾아낼 수 있습니다. 또한 클러스터 수를 사전에 지정하지 않아도 된다는 장점도 있습니다.


# 과제2

 문제3) 아래 코드의 빈칸을 채워 PCA와 LDA 차원축소 과정을 완성하고 실행 결과를 확인하세요

```python
import matplotlib.pyplot as plt

from sklearn import datasets
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

### 데이터 로드

```python
iris = datasets.load_iris()

X = iris.data
y = iris.target
target_names = iris.target_names

print("데이터 shape:", X.shape)
print("클래스:", target_names)
```

### PCA 적용

```python
pca = PCA(n_components=2)   #  PCA를 2차원으로 설정
X_pca = pca.fit(X).transform(X)

print("PCA explained variance ratio:", pca.explained_variance_ratio_)
```

### LDA 적용

```python
lda = LinearDiscriminantAnalysis(n_components=2)  # LDA를 2차원으로 설정
X_lda = lda.fit(X, y).transform(X)  #  라벨 포함하여 학습 후 변환
```

### PCA 결과 시각화

```python
plt.figure()
colors = ["navy", "turquoise", "darkorange"]
lw = 2

for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    plt.scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        color=color,
        alpha=0.8,
        lw=lw,
        label=target_name,
    )

plt.legend(loc="best")
plt.title("PCA of IRIS dataset")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

### LDA 결과 시각화

```python
plt.figure()

for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    plt.scatter(
        X_lda[y == i, 0],
        X_lda[y == i, 1],
        alpha=0.8,
        color=color,
        label=target_name,
    )

plt.legend(loc="best")
plt.title("LDA of IRIS dataset")
plt.xlabel("LD1")
plt.ylabel("LD2")
plt.show()
```

문제4) 코드를 실행했을 때  LDA 시각화가 PCA 시각화보다 클래스 분리가 더 잘 나타나는 이유를 설명하세요.

답변:
- LDA 시각화가 PCA보다 클래스 분리가 더 잘 나타나는 이유는, 두 알고리즘의 목적 함수 자체가 다르기 때문입니다.

- PCA의 한계:
PCA는 클래스 label을 전혀 고려하지 않고, 오직 데이터 전체의 분산이 최대화되는 방향으로 축을 찾습니다. 따라서 분산이 크다고 해서 반드시 클래스가 잘 분리된다는 보장이 없습니다. 즉, PCA가 찾은 주성분 방향이 클래스를 잘 구분하는 방향과 일치하지 않을 수 있습니다.

- LDA가 더 잘 분리되는 이유:
반면 LDA는 클래스 label을 직접 활용하여, 클래스 간 분산은 최대화하고 클래스 내 분산은 최소화하는 방향의 축을 찾습니다. 즉, 처음부터 "클래스를 얼마나 잘 분리할 수 있는가"를 목적으로 차원 축소를 수행하기 때문에, 저차원으로 투영했을 때 클래스 간 경계가 훨씬 명확하게 나타납니다.

- Iris 데이터 관점에서:
Iris 데이터는 3개의 클래스(setosa, versicolor, virginica)가 존재하며, 각 클래스가 어느 정도 정규분포를 따르는 구조입니다. 이는 강의 자료에서 언급한 "LDA가 더 좋은 성능을 내는 조건" — 데이터가 정규분포를 따르고 클래스별 공분산이 유사한 경우 — 에 부합하기 때문에 LDA가 특히 효과적입니다.
